# VoiceGuard AI — Voice Cloner (demo tool, XTTS-v2)

**This is a SEPARATE demo tool — not part of the VoiceGuard app.** It makes a fake
AI clone of a voice so you can test the detector.

**How to use:** open in Google Colab → **Runtime → Change runtime type → T4 GPU** →
run the cells top to bottom → download `cloned.wav` → upload it into the VoiceGuard app.

⚠️ Clone only a voice you have **consent** for (e.g. your own family member, for this demo).

```python
# 1) Install the cloner + a COMPATIBLE transformers.
#    coqui-tts 0.27.x needs transformers>=4.57, but transformers 5.x removed a function it uses
#    (isin_mps_friendly) -> so pin to the 4.57..4.x range (>=4.57 and <5).
!pip install -q coqui-tts "transformers>=4.57,<5"
import os
os.environ["COQUI_TOS_AGREED"] = "1"   # non-commercial model licence, fine for a demo
print("Installed. IMPORTANT -> Runtime -> Restart session, then run from Cell 2 (skip this cell).")
```

```python
# 2) Check the GPU is on (if False: Runtime -> Change runtime type -> T4 GPU, then re-run)
import torch
print("GPU available:", torch.cuda.is_available())
if not torch.cuda.is_available():
    print("⚠️ No GPU — it will be very slow. Turn on the T4 GPU and re-run.")
```

```python
# 3) Upload a clean 6–15 second clip of the REAL person's voice (wav / mp3 / m4a)
from google.colab import files
print("Choose a clean clip of the REAL person's voice:")
up = files.upload()
REF = list(up.keys())[0]
print("reference clip:", REF)
```

```python
# 4) Clone the voice. Edit TEXT (what the fake should say) and LANG, then run.
from TTS.api import TTS
import torch

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("loading XTTS-v2 (downloads ~2GB the first time)...")
tts = TTS("tts_models/multilingual/multi-dataset/xtts_v2").to(DEVICE)

# ---------- EDIT THESE ----------
TEXT = "बेटा, मुझे अभी तुरंत पैसे चाहिए। जल्दी से OTP बताओ।"
LANG = "hi"   # "hi" = Hindi (use Devanagari text) , "en" = English
# --------------------------------

tts.tts_to_file(text=TEXT, speaker_wav=REF, language=LANG, file_path="cloned.wav")
print("✅ done -> cloned.wav")
```

```python
# 5) Listen to it here before downloading (check it sounds like the person)
from IPython.display import Audio
Audio("cloned.wav")
```

```python
# 6) Download cloned.wav to your computer, then upload it into the VoiceGuard app
from google.colab import files
files.download("cloned.wav")
```
